## 클래스

### 1. 객체지향과 클래스, 인스턴스
- 객체(object) = 상태(데이터,변수), 행동(메서드,함수)을 가진 것
- 객체 지향 프로그래밍 = 객체간의 상호작용으로 이루어지는 프로그래밍

```python
class User:
    def __init__(self, name):
        self.name = name # 상태(데이터)

    def show(self):
        print(f"회원 이름: {self.name}") #행동(메서드)

user = User('철수')

user.show()
```

- __클래스__ : 객체를 만들기 위해 속성과 행동을 미리 정의 해둔 '설계도'
- __인스턴스__ : 설계도(클래스)를 바탕으로 메모리에 실제로 만들어진 '객체'

- 차 = 휠, 차체, 엔진... 등등의 구성과 기능을 가짐
- 차1 = xx휠, xx차체, xx엔진... 차2 = yy휠, yy차체, yy엔진... 처럼 기본 설계도(틀)를 기준으로 다양한 차를 생산

### 2. 생성자와 self 키워드
- \_\_ini\_\_ : 메서드를 통해 객체가 생성될때 가장 먼저 호출하는 함수
    - 주로 데이터를 선언하며 최초의 값을 정의 한다
- self : 객체 자기 자신을 가리키는 참조 변수
    - 파이썬 에서는 클래스 내부 함수에서는 첫 번째 매개변수로 반드시 self(객체 매개 변수)를 지정해야 한다. (그래야 파이썬이 알아 들을 수 있음)
    - 다만 반드시 'self' 일 필요는 없지만 강력한 관례로 다른 작업자의 작업물을 보더라도 쉽게 파악할 수 있으므로 self 를 쓴다

```python
class Test:
    def __init__(self, name):
        self.name = name # 값 정의
        self.pp = 100 # 기본값

    def talk(asd):  #self가 아니어도 작동은 한다
        print(f"내 이름은 {asd.name}")

tt = Test('철수')

tt.talk()
```

```python
# __str__ 과 __repr__

#__str__

class Test:
    def __init__(self, name):
        self.name = name # 값 정의
        self.pp = 100 # 기본값

    def talk(self):  
        print(f"내 이름은 {self.name}")

    def __str__(self):
        return "print(객체) 시 보여지는 문장" #print 해서 보여주므로 str을 반환하면 된다

ttt = Test('철희')
print(ttt)

#__repr__

class Test:
    def __init__(self, name):
        self.name = name # 값 정의
        self.pp = 100 # 기본값

    def talk(self):  
        print(f"내 이름은 {self.name}")

    def __repr__(self):
        return f"name={self.name}" #print 해서 보여주므로 str을 반환하면 된다

tt2 = Test('영희')
print(tt2)

# __str__ 과 __repr__ 을 같이 사용하고 print 하는 경우 __str__을 우선 한다
# str은 사용자에게 보여줄 문자열, repr은 개발자에게 보여줄 객체를 설명하는 문자열
```

## 상속과 캡슐화

- 상속 : 이미 만들어진 클래스(부모 클래스)의 속성과 메서드를 그대로 물려받아 새로운 클래스(자식 클래스)를 정의 하는 기술
- 메서드 오버라이딩 : 부모 클래스에 정의된 메서드를 자식 클래스에서 새롭게 정의 하는것
    - 자식 클래스에서 부모 클래스와 동일한 메서드명으로 선언하면 된다
    - 자식 객체에서 메서드를 호출 시 부모의 원본 메서드 대신 자식의 메서드가 덮어쓰면서(overriding) 하면서 자식 메서드가 실행 된다.
- super() : 자식 클래스 내부에서 부모 클래스의 메서드나 상태를 호출할 때 사용
    - 자식 클래스의 새로운 속성을 추가하면서 부모 클래스의 기존 초기화 로직을 재사용하거나 코드 중복 없이 부보 클래스의 자원을 안전하게 이어받을 때 사용

```python
class Person:
    def __init__(self, name):
        self.name = name

    def over_test(self):
        print(f"나는 {self.name}")

class Student(Person):
    def __init__(self, name, major):  # super()를 통해 부모 클래스의 생성자를 실행하여 name을 먼저 세팅
        super().__init__(name)
        self.major = major  # 학생만의 고유 속성 추가
    
    def over_test(self):
        print(f"나는 학생 {self.name}")

std = Student("김철수", "컴퓨터공학")
print(std.name)   # 부모에게 물려받은 속성이 잘 채워졌는지 확인합니다.
print(std.major)  # 자식 고유의 속성도 함께 확인합니다.

std.over_test()   # 자식 클래스의 over_test 실행 확인
```
