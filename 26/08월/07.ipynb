    - 논리형 : True, False
    - 문자형 : "" 감싸서 선언
    - 값 없음 : None

```python
"""
명시적 형변환
int() : 정수로 변환
float() : 실수로 변환
bool() : True,False로 변환
str() : 문자형으로 변환
"""

num = "100"
print(type(num))

print(type(int(num)))

# 묵시적 형변환
num1 = 100
num2 = 3.14

num3 = num1 + num2

print(num3)
```

```python
# 거짓으로 인식하는 값
# 0, "", [], None

message = "1"
if message: #값이 1 -> 참 -> print 실행
    print("있음")

message = ""
if message: #값이 "" -> 거짓 -> print 실행 x
    print("있음")
```

## 연산자
- 대입 연산자(=)
- 산술 연산자(+,-,/,//,*,**,%)
- 비교 연산자 (<,>,<=,>=,==,!=)

```python
# 동등과 동일의 차이

num1 = 1000
num2 = 1000

#동등
print(num1==num2) # 같은 1000 이므로 True 출력
print(num1 is num2) # 주소가 다르므로 False 출력 -> is는 주소 비교
```

```python
# 단락 회로 평가
# and, or 연산자는 앞 항의 결과에 따라 뒷 항의 연산을 안 할 수 있다
# and - 앞 항이 false 인 경우 , or - 앞 항이 true 인 경우
# 앞 항에서 이미 결과를 만족 했으니 안함

print(False and 500//0) # 뒷항이 제로디바인 에러지만 연산을 하지 않아 false를 출력하고 정상 실행
print(True or 500//0) # 마찬가지
```

##### 연산 우선 순위
- 대입 연산자(=)  <  논리 연산자  <  비교 연산자 < 산술 연산자 < (..)

```python
# 문자열 대체
# fstring 방식만

menu = "Americano"
price = 3000

print(f"메뉴 : {menu}, 가격 : {price:>5,}")

# 앞에 f 붙이고 {} 안에 변수명, 변수명 뒤에 : 입력으로 옵션 추가 가능 >,< 숫자(n칸 확보) ','옵션은 숫자에 ,
```
