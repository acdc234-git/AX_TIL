## 함수 정의와 매개변수
- 함수 : 매개변수 입력 -> 함수 처리 -> 값 반환

- 컴파일 기반
    - 소스코드 -> 컴파일 -> 기계어 -> 실행
    - C, C++, JAVA...
    
- 인터프리터 기반
    - 소스코드 -> 한줄씩 해석(실행중) -> 결과
    - 파이썬, 자바스크립트 ...

### 함수의 작동 정리
- 파이썬에서의 함수는 '객체'이다
- def func() 등으로 함수를 선언시 동적 메모리의 힙영역에 객체로 저장 된다
- 따라서 var_func = func 등으로 변수를 함수로 선언할 수 있다
- 함수는 호출시 Call Stack에 로드되며 작동 완료(return 실행)후 제거 된다


#### 매개변수와 인수
- 여타 언어와 비슷
- def func(매개변수) 로 선언
- func(인수) 로 호출
- 위치 인수
    - def func(변수1, 변수2) 등으로 여러 매개 변수를 선언
    - func(변수1,변수2) 로 입력시 인수가 순서대로 매개변수에 대입
- 키워드 인수
    - def func(변수1, 변수2) 동일하게 선언
    - func(변수2=1, 변수1=2) 로 (매개변수=값) 으로 선언 시 위치와 상관없이 값이 대입된다
- 가변 인수
    - def func(*변수) 변수 앞에 * 을 붙여 선언
    - func(변수1,변수2,변수3,...) 등으로 여러개를 넣을 수 있다.
    - def func(변수1,변수2,*변수) 가변변수는 일반 매개변수보다 뒤에 선언 되어야 한다
- 키워드 가변 인수
    - 인수의 이름을 지정하여 키,값의 딕셔너리 구조로 전달 되는 매개 변수
    - def func(**변수) 변수 앞에 ** 을 붙여 선언
    - func(키1=값1, 키2=값2) 등으로 키=값 형태로 인수를 입력하여 호출한다
- return
    - return 값
    - 실행시 값을 호출한 코드에 반환하고 함수는 종료된다
    - return 값, 값2 등으로 복수 반화 할 수 있다 (튜플 구조로 전달)
    - 패킹, 언팩을 활용 할 수 있다
        - value1, value2 = 함수 -> 함수의 반환값 2개를 value1, value2 에 나눠 받는다
        - value1 = 함수 -> value에 튜플구조로 받는다
        - value1, *value2 = 함수 -> value 에 하나의 값 value2 에 나머지 값들을 튜플로 받는다


```python

# 두 개의 숫자를 받아서 두 숫자의 곱을 반환하는 함수
def multiply(x,y):
    return x*y
print(multiply(4,6))

# 사용자의 이름과 나이를 받아 다음과 같이 출력하는 함수 제작
# 이름 : 철수, 나이 : 20 나이는 기본값 20으로 설정

def introduce(name,age=20):
    print(f'이름: {name}, 나이: {age}')

introduce('철수')
introduce('영희',25)
introduce(age=30,name='민수')
```

```python
def print_user(**args):
    for key, value in args.items():
        print(f"{key}: {value}")

print_user(
    name="철수",
    age=25,
    job="개발자",
    level=3
)
```

```python
# 숫자를 입력받아 총합과 평균을 구하는 함수

def calculate(*num):
    if len(num) == 0:
        return 0,0
    else:
        return sum(num), sum(num)/len(num)

result = calculate(10,20,30,40)
print(result)
```
