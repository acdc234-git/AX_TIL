### datetime

- import datetime 으로 사용 할 수 있다
- datetime.datetime.now() : 현재 날짜와 시간(시,분,초) 를 반환하는 함수
- datetime.date.today() : 현재 날짜 정보(연,월,일)만 반환
- datetime.timedelta() : 두 날짜나 시간 사이의 간격을 표현하거나 계산
- time.sleep(secs) : 인자로 전달된 초(second) 만큼 프로그램의 실행을 일시 정지
- time.perf_counter() : 소요 시간 측정에 적합한 높은 해상도의 고정밀 타이머 값을 반환
- datetime.strftime() : 시간을 형식화 표현
    - %Y	4자리 연도	2026
    - %m	2자리 월 (01 ~ 12)	06
    - %d	2자리 일 (01 ~ 31)	30
    - %H	24시간 형식의 시 (00 ~ 23)	14
    - %M	2자리 분 (00 ~ 59)	45
    - %S	2자리 초 (00 ~ 59)	12
    - %A	요일 이름 (영어 기준 full name)	Tuesday
    - 사용 예시 -> now.strftime("%Y년 %m월 %d일 %H시 %M분 %S초)
    

### json
- json.dumps(obj) : 파이썬 객체(딕셔너리, 리스트)를 JSON 형식의 문자열로 변환
- json.loads(s) : JSON 형식의 문자열을 파이썬 객체로 역변환
- json.dump(obj, fp) : 파이썬 객체를 JSON 형식의 파일로 직접 저장할 때 사용
- json.load(fp) : 저장된 JSON 파일을 읽어와 파이썬 객체로 역변환

- dumps -> 객체를 __문자열__ 로, dump -> 객체를 __파일__ 로

```python
import json

#파이썬의 딕셔너리 데이터
py_dict = {
    '예시':1,
    'json':2,
    '데이터':3
}

#파이썬 데이터를 JSON 문자열로 (직렬환)
json_text = json.dumps(py_dict,ensure_ascii=False,indent=4) #indent 는 각 데이터를 줄바꿈 하면서 들여쓰기를 어느정도 할건지 4 = 4칸

print(json_text)
type(json_text) #str로 직렬화 되었다

re_dict = json.loads(json_text) # 다시 파이썬 딕셔너리로

print(re_dict)
type(re_dict)

# dumps loads -> 뒤에 s 가 붙은것은 문자열인 str의 약자. dumps 문자열로 저장 loads 문자열을 불러옴. s가 없으면 파일을 불러와야 되므로 with open 의 형식이 필요하다
```

## 동기와 비동기

- 동기 : 앞선 작업부터 하나씩 완료한다
- 비동기 : 이전 작업의 완려 여부와 상관없이 각 작업을 병렬로 진행한다
    - 비동기로 작업을 위해선 async 키워드가 필요하다
        - 동기 함수 : def 함수명()
        - 비동기 함수 : async def 함수명()
            - await 키워드를 통해 해당 함수를 대기 시키고 그 동안 다른 작업을 진행한다
