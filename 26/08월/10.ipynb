## 자료구조
    - 데이터 군집 -> collection
    - C(creat) : 생성
    - R(read) : 조희
    - U(update) : 수정
    - D(delete) : 삭제

1. 순차 자료구조
    - 순서가 있는 데이터

2. 집합 자료구조
    - 중복을 허용하지 않는다
    - 순서를 결정하지 않는다 (1,2,3,4 = 2,3,4,1)

3. 사전 자료구조
    - 키워드 - 찾은 값
    - key - value 쌍 구조
    - key는 중복을 허용하지 않는다 (집합의 특성)
    - value는 중복을 허용한다

```python
# list
ex_list = ['1',2,3]
print(ex_list[0])
print(type(ex_list[0]))
print(ex_list[1])
print(type(ex_list[1]))
print(ex_list[2])
print(type(ex_list[2]))
```

```python
#tuple
ex_tuple = (1,2,'3')
ex_tuple2 = 1,2

print(ex_tuple)
print(ex_tuple2)
```

- 인덱싱과 슬라이싱
- list 와 tuple 같은 시퀀스(순차) 자료구조는 번호(인덱스)를 통해 값을 찾거나 범위 만큼 잘라낼 수 있다

```python
# 인덱싱
ex_list[0]
ex_list[-1]

#슬라이싱
ex_list[0:3]
ex_list[:2]
ex_list[0:5:2] # 세번째는 증감폭
ex_list[::-1] # 역순 정렬
```

```python
#길이 측정과 포함 여부 
len(ex_list) # ex_list의 길이 측정
print(1 in ex_list) # 1 값이 ex_list에 있는가 확인
print('1' not in ex_list) # '1' 값이 ex_list에 없는지 확인

#패킹

packing = 1,2,3,4 # 괄호 없이 나열하면 자동으로 튜플로 패킹
a,b,c,*d = packing # 변수 a,b,c,d에 패킹된 값들을 순서대로 나눠서 저장, 변수명 앞에 *이 붙는경우 나머지 값들을 해당 변수에 넣어서 패킹
                   # *에 패킹되는 자료는 list 형이다

```

```python
# key-value 딕셔너리 자료구조
ex_dict = {'key':'value',1:2, '1':2}

print(ex_dict.keys())
print(ex_dict.values())
print(ex_dict.items())

print(ex_dict[1])
print(ex_dict.get(3,'없는 키'))
```

```python
# 중복 없는 자료구조 set

ex_set = [4,1,1,2,3,4,4]
set(ex_set)
print(set(ex_set))
print(ex_set) # 새로 만드는것이 아니기에 set을 벗어나면 돌아간다

tmp_set = set(ex_set)
print(tmp_set) #set() 함수의 결과를 다른 변수에 저장

list_set = list(tmp_set) #set()을 다시 list 구조로 돌려서 저장 -> list 구조인 경우 검색과 순서대로 정렬등이 가능하기 때문에

```
