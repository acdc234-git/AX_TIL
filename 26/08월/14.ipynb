## 이터러블과 이터레이터
- 반복 가능한 객체 이터러블(Iterable)
- 실제 순회를 제어하는 이터레이터(Iterator)

### 이터러블
- 리스트, 문자열, 딕셔너리 처럼 내부 요소를 하나씩 꺼내어 순회 가능한 모든 객체
- 내부적으로 __iter__ 메서드를 가지고 이터레이터를 생성

### 이터레이터
- 이터러블 객체에서 요소를 순서대로 꺼내는 포인터 객체
- __next__ 가 다음 요소를 가리키고 있다
- 한번에 하나씩 데이터를 메모리에 불러와 처리

```python
#iter() 함수로 이터레이터 객체를 만들고 next() 로 __next__를 실행하여 데이터를 하나씩 꺼내기

#이터러블 객체 생성
fruits = ['사과','바나나','체리']

#이터레이터 객체
fruits_iterator = iter(fruits)

print(next(fruits_iterator))
print(next(fruits_iterator))
print(next(fruits_iterator))
# 다음 데이터가 없을때 꺼내면 에러 발생
```

### 제네레이터(Generator) 함수
- yield 까지 함수를 실행하고 yield를 만나면 그때까지의 결과값을 반환하고 일시 정지 상태가 되는 함수
- 일시 정지 동안은 메모리에 올라간 상태로 대기한다
- next()를 통해 다시 요청이 오는 경우 다음 yield 까지 실행한다

```python
#지연 평가 (Lazy Evaluation)

def infinite_number():
    number = 1  #최초 1로 초기화
    while True: #무한 루프 진입
        yield number #최초 1 반환 후 정지, 증가된 값 반환후 다시 정지
        number += 1  #1 증가 후 루프 실행
# -> next() 를 실행할때 마다 1 증가된 값을 반환하고 멈춘다

#제네레이터 객체 생성
num_gen = infinite_number()
#infinite_number() 만 가지고 실행하게 되면 계속 새로운 객체를 생성하므로 1만 출력 된다
#따라서 현재 상태를 기억할 수 있게 변수에 infinite_number() 을 넘긴다

print(num_gen) #<generator object infinite_number at 0x0000022F5D30B640>
print(infinite_number)#<function infinite_number at 0x0000022F5D5C1E40>
#num_gen 은 제네레이터 객체이고 #infinite_number은 함수로 다른걸 알 수 있다

print(next(num_gen)) #1
print(next(infinite_number())) #1
print(next(num_gen)) #2
print(next(infinite_number())) #1
print(next(num_gen)) #3
print(next(infinite_number())) #1
```

```python
#제네레이터 표현식과 리스트 컴프리헨션

m_list = [num for num in range(10000)] #리스트 컴프리헨션
g_list = (num for num in range(10000)) #제네레이터 표현식
#대괄호와 소괄호로 나뉜다
import sys

print(sys.getsizeof(m_list)) #생성하게 되는 모든 데이터 메모리를 생성 한다
print(sys.getsizeof(g_list)) #요청이 오면 수행하므로 코드 메모리만 보유하고 있다
```

## 모듈과 패키지
### 모듈
- 파이썬의 모듈은 하나의 .py 파일이다
- 모듈을 기능별로 나눠 저장하여 작업 효율성을 높일 수 있다
### 패키지
- 여러 모듈(.py)를 하나의 폴더에 담는걸 패키지라고 한다
- 패키지 내부의 패키지 혹은 패키지 상위의 패키지 등 여러 구조가 존재 할 수 있다
- 이러한 계층 구조를 현재 폴더(.), 상위 폴더(..) 등으로 접근할 수 있다

```python
#모듈 로드를 위한 키워드

import math #특정 모듈 전체를 현재 파일의 메모리에 로드 모듈이름.함수이름() 형태로 호출한다
math.sqrt(16) #모듈이름.함수

from math import sqrt
sqrt(16) #모듈명 없이 함수이름만으로 호출

import math as m
m.sqrt #math 를 별칭 m으로 변경하여 호출한다
```
