```python
# 동기, 비동기 예제 코드

import asyncio #비동기 함수 사용을 위한 라이브러리
import time

async def ex_async(tesk,seconds): #async def 형태로 함수를 선언
    print(f"{tesk} 시작 ")
    await asyncio.sleep(seconds) # sleep 가 끝나길 기다려주는 await
    print(f"{tesk} 종료 ")

async def main(): # 비동기를 위해 main도 async 선언
    start = time.time()
    await asyncio.gather(ex_async('첫번째', 2),ex_async('두번째',1)) #await 를 안 쓰면 안 기다리고 다음라인을 실행 한다
    end = time.time()

    print("소요 시간: ",round((end-start),2))

await main()

# asyncio.gather 로 2개의 함수를 동시에 실행하고 각각 끝날때 까지 대기후 모두 끝나면 같이 결과를 반환하며 다음 줄을 실행한다
```

## AI 기본 개념
- AI > 머신러닝 > 딥러닝 > 생성형 AI(LLM)
- 전통적 AI = if-then 으로 이어지는 사람의 하드 코딩
- 머신러닝 = 데이터를 학습하고 패턴을 찾아내어 스스로 정답을 찾아내게 함

### LLM과 트랜스포머 기술
- LLM : 방대한 텍스트 데이터를 흡수해 단어들 사이의 통계적 관계(임베딩 벡터)를 학습하여 사용자가 전달한 프롬프트 뒤에 '통계적'으로 가장 자연스럽게 이어질 다음 단어를 '예측' 하여 보여주는 방식
- 트랜스포머의 셀프 어텐션 : 전체 묵맥 속 유기적인 관계 가중치를 계한해내며 여러 대의 컴퓨터가 동시에 작업을 나누어 처리하는 병렬 연산을 가능하게 함

## AI 리터러시 (AI 문해력)
- 확률적 제안에 대한 인지 : '통계적'으로 다음 토큰에 올 확률이 가장 높은 단어를 선정하므로 AI는 '생각'을 하는 것이 아닌 '계산'을 하고 있는것이다. 방대한 텍스트의 학습으로 인한 통계적으로 그럴듯해 보이는 단어들을 나열하여 문장을 만드는것을 항상 인지 해야 한다.
- 환각 현상에 대한 경계 : 따라서 통계적으로 확률이 높은 단어들을 나열하는 것이므로 잘못된, 허황된 결과가 나오는 경우가 있으므로 항상 결과물에 대해 교차 검증을 해야 한다

### 토큰화와 임베딩
- 문장을 잘라낸 '토큰화' 와 잘라낸 단어를 숫자 벡터로 변환하는 임베딩을 통해 컴퓨터는 비로소 문장을 읽을 수 있게 된다.
- 토큰화 : AI 모델이 텍스트를 읽고 쓰는 가장 기본적인 최소 식별 단위
    - 하나의 단어 = 하나의 토큰 이 아니다
    - 여어는 대개 한 단어가 1개의 토큰으로 묶이는 경우가 많다
    - 한국어는 조사와 어미가 발달한 교착어이므로 하나의 단어라도 더 작은 글자 단위로 쪼개져 2~3개의 토큰으로 분할 되기도 한다

- 임베딩 : 토큰화 과정을 통해 잘게 쪼개진 토큰들을 컴퓨터가 연산할 수 있도록 실수로 이루어진 다차원 벡터 공간의 좌표로 치환하는 과정
    - 공간 정리 기법 : 기계적인 단어 나열을 넘어 단어가 가진 주제, 맥락, 분위기 등 수백 가지의 평가 기준을 바탕으로 공간을 정리한다
    - 패턴 분식 및 학습 : 방대한 텍스트 데이터를 사전 학습하여 문장 속에서 단어들이 가지는 숨겨진 패턴과 맥락을 끊임없이 분석한다
    - 의미적 유사성 매핑 : 의미적 유사성을 가진 단어일수록 임베딩 공간 속에서 서로 가까운 거리에 배치되며 맥락이 전혀 다른 단어는 먼 거리에 위치하게 된다

### 확률 기반 다음 단어 예측
- 추론 과정
    1. 신경망 통과 : 사용자가 입력한 프롬프트 문맥을 인식하고 학습 데이터 기반의 단어 사전을 가상으로 전개
    2. 확률 계산 : 주어진 문맥 뒤에 위치할 가능성이 있는 후보 단어들의 통계적 확률을 정밀하게 계산
    3. 토큰 출력 : 계산된 지표 중 압도적으로 높은 확률을 획득한 단어를 최종 선택하여 화면에 노출
    4. 문장 업데이트 및 반복 : 새롭게 출력된 단어를 기존 문장에 포함하여 다시 전체 입력값으로 집어넣고 그 다음 토큰을 또 다시 새롭게 예측
    5. 프로세스 종료 : 지정된 분량이 모두 채워지거나 모델 스스로 문장의 완전한 끝을 판별하는 특수 토큰(EOS, End Of Sequence)을 도출할 때까지 이 순환 구조를 초당 수십 번씩 반복

- 창의성을 조절 하는 Temperature
    - 온도의 역할 : 인공 지능이 상황에 따라 확률 분포 내의 2등, 3등 토큰도 선택할 수 있도록 무작위성과 창의성 비율을 제어하는 매개변수

- 구조적 부작용 환각(Hallucination)
    - 고정된 데이트베이스에서 검색 하여 가져오는 시스템이 아님
    - 매 순간 통계적으로 가장 자연스러운 문장을 생성 해내는 구조이기 때문에 사실 검증을 하지 않음
    - 따라서 정보가 부족한 상황에서도 문맥상 그럴듯해 보이는 문장을 만들어냄

### 컨텍트스 윈도우 (Context Window)
   - AI 모델이 한번의 추론 과정에서 동시에 읽어 들이고 기억하며 참고할 수 있는 텍스트(토크)의 최대 허용 길이

- 히스토리 처리 : 새로운 답변을 추론할때 이전의 대화 기록 전체를 한데 묶어 입력값으로 처리
- 과거 데이터의 삭제 : 대화가 길어져 토큰 총량이 컨텍스트 윈도우의 제한 용량을 넘어서면 가장 오래된 토큰 부터 삭제
- 기억 상실 : 초반부에 지정한 프롬프트들이 유실될 수 있으며 과거 대화를 망각할 수 있다
