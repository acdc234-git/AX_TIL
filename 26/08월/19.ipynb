## 예외 처리

- 구문 오류 (Syntax Error) : 문법을 잘못 적음
    - ex) 오타, 들여쓰기 실수, 괄호 누락
- 런타임 예외(Exception) : 잘못된 연산을 수행 함
    - ex) 0 나누기, int+str, 리스트를 벗어난 인덱스 값 접근

- 구문 오류는 실행전에 알 수 있으나 런타임 에러는 실행해봐야 알 수 있다

```python
#IndexError 
nums = [1,2,3,4,5]
print(nums[9]) # 데이터는 0~4 까지 있으나 데이터가 없는 9번에 접근 -> IndexError 출력

#ValueError
text = '텍스트'
num = int(text) # text는 문자열로 int형 숫자로 변환이 불가

#TypeError
result = 10 + '10' # int + str

#ZeroDivisionError
result = 10 / 0 # 0으로 나누기는 불가능 하므로
```

```python
# try-except-else-finally
try:
    value = int('123')
except ValueError:
    print('예외 발생시 실행')
else:
    print('예외가 없으면 실행')
finally:
    print('예외 발생 여부 상관 없이 실행')
```

```python
# 예외를 발생시키는 raise

def test_raise(num):
    if num >= 10:
        # num이 10보다 큰 수가 들어와도 구문 오류나 런타임 예외가 발생하지 않으나 원하지 않는 값이 들어왔기에 예외 구문으로 던진다
        raise ValueError('10보다 큰 값은 들어올 수 없다')
    return num

try:
    test_raise(10)

except ValueError as e:
    print(e)
```

### raise의 개념 이해
- 런타임 예외가 발생하지 않는 문법이어도 예외를 발생시킨다
- 그러므로 개발자가 원하지 않는 처리가 되었을때 예외 블록에서 처리 하는 방식을 정해서 개발할 수 있다
    - if-elif-else 구문으로도 개발이 가능하나 너무 길어지고 보기 힘들다

```python
def set_user_age(age):
    if age < 0:
        # 의도하지 않는 값이 들어오는 경우 raise로 예외에 던진다
        raise ValueError("나이는 음수가 될 수 없습니다.")
    print(f"사용자의 나이가 {age}세로 설정되었습니다.")

try:  # try는 정의 하는 함수가 아닌 선언하는 블록에 들어간다
    input_age = -5
    set_user_age(input_age)
except ValueError as e:
    # 발생한 예외를 식별하여 메시지 출력
    print(f"[시스템 경고] {e}")
```

## 파이썬의 내장 함수

```python
# isinstance() 자료형 검증 -> 변수와 입력한 자료형을 비교하고 True, False를 반환한다

age = 10
ages = '10'

print(isinstance(age,int)) # int 가 맞으므로 True
print(isinstance(ages,int)) # int가 아니므로 False
```

```python
#all(), any()

#all() - 목록의 데이터가 전부 True 면 True 반환

agree = [True,True,True]
all(agree) # True

no_agree = [True,False,True]
all(no_agree) # 하나가 False 이므로 False

any(no_agree) # 하나라도 true 이므로 true
```
