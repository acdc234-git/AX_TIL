### 모델 파라미터와 하이퍼 파라미터
- 모델 파라미터 : 인공지능이 훈련 과정을 토해 스스로 학습하고 결정하는 값
    - 가중치(w), 편향(b)

- 하이퍼 파라미터 : 인간이 학습을 시작하기 전에 미리 세팅해 주어야 하는 수동 조절 다이얼
    - 학습률, 학습 횟수(Epochs)등

### 데이터 분할 (Data Split)
- 모든 데이터를 한 번에 주고 학습 시키게 되면 질문과 답을 외워버리는 과적합 현상이 발생함
- 전체 데이터를 Scikit-learn 의 train_test_split 기능을 활용하여 3개로 분리
    - 훈련 세트 (60%) : 인공지능이 학습할 데이터
    - 검증 세트 (20%) : 학습에 대한 평가용 데이터
    - 테스트 세트 (20%) : 학습이 끝난 후 모델이 본 적없는 데이터를 입력하여 출력을 테스트

### 데이터 스케일링
- MinMaxScaler : 가장 큰 값을 1 가장 작은값을 0으로 맞춰 데이터를 0~1 사이로 조정
- StandardScaler : 데이터의 평균을 0점으로 표준 편차를 1로 맞추어 데이터가 평균으로 부터 얼마나 떨어져 있는지 표준 점수로 조정.(평균에서 1만큼 떨어진 데이터다)



### 다중 선형 회귀
- 여러개의 원인을 입체적으로 조합하여 하나의 결과를 예측하는 알고리즘

#### y = w1x1 + w2x2 + w3x3 + ... + wpxp + b

- y : 예측값
- x1,x2,x3... : 각 변수들 (예측값의 원인들.. 집 크기, 범죄율, 학군 등등)
- w1,w2,w3... : 각 변수들에 대한 가중치
- b : 편향. 기본적인 값

### 다중 공선성
- 원인 변수들끼리 서로 지나치게 밀접한 상관 관계를 가져서 독립적인 기여도를 파악하기 어려워지는 중복 간선 현상
- 유해성 : 데이터가 조금만 바뀌어도 중요도 평가(가중치 계수)에 변동이 크게 오게되어 모델의 신뢰도가 크게 추락
    - 모델의 과적합 위험을 크게 키워 새로운 데이터를 마주했을 때 예측력이 상실

### 특성 공학과 규제 모델
- 특성 공학 : 기존의 변수를 가공해서 더 유용한 다서를 새롭게 만들어내는 __전처리 과정__ . 변수가 증가하므로 과적합 부작용이 발생할 수 있으므로 변수에 페널티를 부여해 과도한 영향력을 억누르는 __규제__ 가 필요

### 릿지 회귀와 L2 규제
- 릿지 회귀 : 예측 평균 오차를 조율하면서 가중치들의 제곱합으로 추가 페널티로 부과해 특정 변수로 쏠리는 현상을 방지

$$
\text{Loss}_{\text{Ridge}} = \text{MSE} + \alpha \sum_{j=1}^{p} w_j^2
$$

- 평균제곱오차 + 규제강도(a) x 모든 독립변수들의 제곱합
- 편향(b)를 규제 하지 않는 것은 b를 규제하게 되면 그래프가 (0,0)에 가까워 지므로 모델의 유연성이 떨어지게 된다

### 라쏘 회귀와 L1 규제
- RIDGE 와 달리 필요 없는 변수를 제외할 수 있다 (가중치를 0으로)

$$
\text{Loss}_{\text{Lasso}} = \text{MSE} + \alpha \sum_{j=1}^{p} |w_j|
$$

- 모든 독립변수 가중치들의 절대값 합
