## 변수 스코프와 람다 함수
### 변수의 유효 범위
- 전역 변수(global) : 코드 전역에서 접근 가능 함수 외부에 선언
- 지역 변수(local) : 함수 내부에서만 접근 가능 함수 내부에 선언
- 함수 내부에 전역 변수와 같은 이름의 지역 변수를 선언 하고 수정 해도 __지역 변수__ 가 수정 된다


````python
v_global = 10 # 전역 변수

def func():
    v_global = 100 # 지역 변수
    print(v_global)

func()
print(v_global) # func의 v_global 은 전역 변수 v_global 에 영향을 주지 않는다
````

변수의 우선 순위 __LEGB__
1. L(Local) : 작동중인 함수 내부
2. E(Enclosing) : 중첩 함수 구조에서 내부 함수를 감싸고 있는 외부 함수의 공간
3. G(Global) : 함수 외부 소스코드 파일 전체 공간
4. B(Built-in) : 파이썬이 태생적으로 내장하고 있는 print(),len() 같은 기본 함수 공간
- 우선순위는 1>2>3>4

### 람다 함수
- 굳이 만들어서 재활용하지 않을 간단한 함수들은 람다 함수로 만들어 사용
- lambda 매개변수 : 처리식 -> return 은 넣지 않아도 작동된다. 매개변수는 , 를 사용하여 여러개 넣을 수 있다(매1,매2,...)

#### map(), filter() 와 람다 함수의 결합
- map(매개변수1,매개변수2) 매개변수1에 람다식을 넣고 매개변수2에 리스트를 넣어 매개변수2의 데이터를 람다식으로 변환가공 한다.
    - 리스트 데이터로 뽑기 위해선 list(map()) 같이 list() 안에 map()를 넣어야 한다
- filter(매개변수1,매개변수2) map()와 동일하게 매개변수를 삽입. 람다식의 조건에 맞는 매개변수2의 값들만 저장한다 (참인 값만 선별하여 저장)
    - 마찬가지로 list() 사용

````python
numbers = [1,2,3,4,5,6,7,8,10]

print(list(map(lambda x: x**2,numbers))) # 리스트의 각 데이터를 제곱하고 리스트에 저장 후 출력
print(list(filter(lambda x: x%2==0, numbers))) #리스트의 각 데이터 중 2의 배수만 저장 후 출력
````

## 클로저와 데코레이터
- 파이썬의 함수는 객체로 사용 가능(일급객체) -> 변수와 마찬가지로 데이터로 사용되어 함수가 종료되어도 메모리에 제거 되지 않는 클로저(Closure) 구조 설계 가능

````python
#일반적인 함수
def func(name):
    return f"이것은 일반 함수 {name} 이다."

#함수를 인수로 전달 받는 고차 함수
def gfunc(func, func_name):
    return func(func_name) # 함수를 실행하고 함수의 결과값을 반환한다

#다른 함수를 결과로 반환하는 고차 함수
def rfunc():
    return func #함수 자체를 반환 !!! 함수를 실행하고 반환하는 것이 아닌 말 그대로 함수 자체를 반환한다 !!!

#함수를 인수로 전달하여 실행 흐름 연결
result_1 = gfunc(func,"고차 함수 예시")
print(result_1) #result_1 에 gfunc 가 인수로 받은 func와 str을 실행해서 result_1에 결과값을 넣는다
type(result_1) #따라서 result_1은 str 이다

#함수를 반환 받아 변수에 매핑 후 작동
my_func = rfunc()
print(my_func("반환")) # my_func는 함수 func를 받은 상태이다.
type(my_func) # 따라서 my_func는 function 이다
````

````python
#클로저는 내부 함수가 외부 함수의 변수를 기억한 채로 반환되어 독립적으로 유지되는 구조이다
#보통 함수가 종료되며 함수 내부의 변수들도 같이 사라지지만 외부의 변수를 참조함으로 없어지지 않는다

def add_num(a):
    def add_again(b):
        return a+b
    return add_again #함수를 반환

add_3 = add_num(3) # a=3 이며 add_again 함수가 된 add_3 함수가 생성

add_3(2) # a=3 인것이 유지된 상태로 add_again 함수의 a+b인 3+2 가 반환

add_10 = add_num(10) # a = 10을 유지하는 함수 add_10

add_10(5) # 10+5
````

### 데코레이터
- 함수에 공통적인 기능 (log 출력, 함수 처리 시간등) 을 붙일때 사용
- 주석처럼 붙여서 사용하여 간단하게 사용 가능하다

````python
def deco(callback):
    def wrapper(*args):
        print("함수시작")
        result = callback(*args)
        print(result)
        print("함수종료")
        return result
    return wrapper

@deco
def add(num1,num2):
    return num2+num1

add(5,10)
````

````python
#데코 함수 정의
def auto_tracker(callback): #callback 함수 매개변수
    def wrapper(*args,**kwargs): # 가변 위치,키워드 인자로 어떤 함수의 매개변수에도 대응
        print(f"[시작] {callback.__name__}")
        print(f"위치 인자(args): {args}")
        print(f"키워드 인자(kwargs): {kwargs}")

        result = callback(*args,**kwargs) # 위에 매개변수로 받은 인자를 인수로 넣고 함수의 결과를 result 에 넣기

        print(f"[종료] {callback.__name__}")
        return result # callback 함수의 결과를 return
    return wrapper

@auto_tracker
def introduce(name, age, job="학생"):
    return f"{name}님은 {age}살이고 직업은 {job}입니다."

result = introduce("철수",20, job='프로그래머')

print(result)
````
