### 데이터 타입
- INT : 정수형 숫자
- VARCHAR(n) : 가변 길이 문자열 (최대 n자)
- DATE : 날짜 (YYYY-MM-DD)
- DECIMAL(p,s) : 소수점이 있는 숫자 (전체 p자리, 소수점 s자리)
- BOOLEAN : 참, 거짓
- TEXT : 긴 문자열

### 스키마
- 스키마(Schema)는 데이터베이스의 구조와 제약 조건을 정의한 설계도
- 어떤 테이블이 있는지, 각 테이블에 어떤 열이 있는지, 각 열의 데이터 타입과 제약 조건은 무엇인지등 그 DB의 명세

### PK 와 FK
- 기본키 : PRIMARY KEY
    - 테이블에서 각 행을 식별하는 열.
    - PK의 규칙
        - 중복 값이 존재할 수 없다
        - NULL 값을 가질 수 없다
- 외래키 : Foreign Key
    - 다른 테이블의 기본키를 참조하는 열
    - FK를 통해 두 테이블이 연결되고 테이블 간의 관계가 형성

### 참조 무결성
- FK 가 참조하는 값이 반드시 참조 대상 테이블의 PK에 존재 해야 한다는 규칙
    - 존재 하지 않는 값을 참조할 수 없다.
    - 따라서 참조 되고 있는 PK는 기본적으로 삭제가 불가능
        - 몇몇 옵션에 따라 삭제 할 수 있다
        - ON DELETE CASCADE : FK 행도 같이 삭제
        - ON DELETE SET NULL : FK 값을 NULL로 변경

### 데이터베이스 정규화
- 데이터 중복을 최소화하고, 데이터 무결성을 높이기 위해 테이블을 분리 하는 과정
- 1규정
    - 하나의 칸에는 하나의 값만 저장
- 2규정
    - 복합키 전체에 종속되도록 분리
- 3규정
    - 일반 컬럼끼리의 의존 관계 제거
