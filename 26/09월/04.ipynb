### 테이블간 관계 유형
- 1:1 관계
	- 한 테이블의 레코드 하나가 다른 테이블의 레코드 하나와 정확히 대응되는 관계
	- 보안이나 성능 목적으로 데이터를 분리할 때 주로 사용

- 1:N 관계
	- 한 테이블의 레코드 하나가 다른 테이블의 여러 레코드와 연결되는 관계
- N:M 관계
	- 양쪽 테이블의 레코드가 서로 여러 개와 연결될 수 있는 관계
	- RDBMS에서 직접 표현이 불가능 하여 중간 테이블을 놓고 1:N - 중간 - N:1 형태로 만들어야 한다

| 기호 | 의미 |
|---|---|
| `\|` (수직선 하나) | 최소 1개, 필수 |
| `O` (원) | 0개 (선택적) |
| `<` (까마귀 발) | 여러 개 (N) |
| `\|\|` (수직선 두 개) | 정확히 1개, 필수 |
| `O<` (원 + 까마귀 발) | 0개 이상 |
| `\|<` (수직선 + 까마귀 발) | 1개 이상 |
